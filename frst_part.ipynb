# Inspection des CSV du Sujet 2

Ce notebook ouvre et présente un aperçu de tous les fichiers CSV de `SujetsHackathon2026/Sujet2/Généralisation`.

Utilise ce notebook pour comprendre rapidement chaque fichier de données, ses colonnes et des exemples de lignes.


```python
import pandas as pd
from IPython.display import Markdown, display
from pathlib import Path

candidate_dirs = [
    Path('.'),
    Path('SujetsHackathon2026/Sujet2/Généralisation'),
    Path('SujetsHackathon2026/Sujet2/Généralisation'),
]
base = next((path for path in candidate_dirs if list(path.glob('*.csv'))), Path('.'))
csv_files = sorted(base.glob('*.csv'))

dataframes = {path.name: pd.read_csv(path) for path in csv_files}

summary = pd.DataFrame(
    [
        {
            'fichier': name,
            'lignes': len(df),
            'colonnes': len(df.columns),
            'noms_colonnes': ', '.join(df.columns),
        }
        for name, df in dataframes.items()
    ]
)

display(Markdown('## Résumé des fichiers CSV'))
display(summary)

for name, df in dataframes.items():
    display(Markdown(f'## {name}'))
    display(Markdown(f'**Dimensions :** {df.shape[0]} lignes × {df.shape[1]} colonnes'))
    display(df.head(20))
```

```python
events = dataframes.get('nodes_events_final.csv')
if events is None:
    events = pd.read_csv(base / 'nodes_events_final.csv')

event_types = (
    events['type']
    .value_counts(dropna=False)
    .rename_axis('type_evenement')
    .reset_index(name='nombre_occurrences')
)

display(Markdown("## Types uniques d'événement - nodes_events_final.csv"))
display(event_types)
```

```python
events = dataframes.get('nodes_events_final.csv')
if events is None:
    events = pd.read_csv(base / 'nodes_events_final.csv')

event_descriptions = (
    events['description']
    .value_counts(dropna=False)
    .rename_axis('description')
    .reset_index(name='nombre_occurrences')
)

display(Markdown("## Toutes les descriptions uniques - nodes_events_final.csv"))
display(Markdown(f"**Descriptions uniques :** {len(event_descriptions)} sur {len(events)} événements"))
with pd.option_context('display.max_rows', None, 'display.max_colwidth', None):
    display(event_descriptions)
```

```python
process_ranking = pd.DataFrame([
    {'rang_processus': 1, 'processus': 'Gestion des Flottes', 'score_criticite_metier': 100, 'impact_si_touche': 'Arrêt ou désorganisation du pilotage flotte, effet direct sur opérations maritimes et disponibilité.'},
    {'rang_processus': 2, 'processus': 'Planification des Itinéraires', 'score_criticite_metier': 95, 'impact_si_touche': 'Risque sur routage, délais, arbitrages opérationnels et continuité de mission.'},
    {'rang_processus': 3, 'processus': 'Suivi des Cargaisons', 'score_criticite_metier': 90, 'impact_si_touche': 'Perte de visibilité sur cargaison, traçabilité et décisions logistiques.'},
    {'rang_processus': 4, 'processus': 'Sécurité et Conformité', 'score_criticite_metier': 88, 'impact_si_touche': 'Dégradation des contrôles, risque réglementaire, sûreté et réponse incident affaiblies.'},
    {'rang_processus': 5, 'processus': 'Douanes et Réglementations', 'score_criticite_metier': 82, 'impact_si_touche': 'Blocage administratif/frontière, retards, non-conformité et immobilisation possible.'},
    {'rang_processus': 6, 'processus': 'Gestion des Équipages', 'score_criticite_metier': 78, 'impact_si_touche': 'Affecte la capacité à armer/opérer les navires et gérer les rotations.'},
    {'rang_processus': 7, 'processus': 'Maintenance des Navires', 'score_criticite_metier': 75, 'impact_si_touche': 'Risque sur disponibilité et sécurité des navires, surtout si panne prolongée.'},
    {'rang_processus': 8, 'processus': 'Gestion des Stocks', 'score_criticite_metier': 68, 'impact_si_touche': "Risque de rupture de pièces, consommables ou marchandises nécessaires à l'exploitation."},
    {'rang_processus': 9, 'processus': 'Approvisionnement', 'score_criticite_metier': 65, 'impact_si_touche': 'Retards fournisseurs et incapacité progressive à soutenir les opérations.'},
    {'rang_processus': 10, 'processus': 'Gestion des Réservations', 'score_criticite_metier': 55, 'impact_si_touche': 'Impact commercial et planification client, critique mais moins vital que opérations flotte.'},
    {'rang_processus': 11, 'processus': 'Gestion des Clients', 'score_criticite_metier': 50, 'impact_si_touche': 'Dégradation relation client et suivi, impact indirect sur opérations.'},
    {'rang_processus': 12, 'processus': 'Finance et Comptabilité', 'score_criticite_metier': 45, 'impact_si_touche': 'Impact financier/reporting, généralement moins immédiat pour la conduite opérationnelle.'},
    {'rang_processus': 13, 'processus': 'Analyse des Risques', 'score_criticite_metier': 40, 'impact_si_touche': "Perte d'aide à la décision et priorisation, grave à moyen terme mais moins immédiat."},
    {'rang_processus': 14, 'processus': 'Communication Interne', 'score_criticite_metier': 35, 'impact_si_touche': 'Coordination dégradée, contournable si canaux alternatifs disponibles.'},
    {'rang_processus': 15, 'processus': 'Ressources Humaines', 'score_criticite_metier': 25, 'impact_si_touche': 'Impact administratif/personnel, rarement le plus critique en incident cyber opérationnel maritime.'},
])

assets = dataframes.get('nodes_assets_it_maritime_10000.csv')
if assets is None:
    assets = pd.read_csv(base / 'nodes_assets_it_maritime_10000.csv')

criticality_weight = {'low': 1, 'medium': 2, 'high': 3, 'critical': 5}
asset_processes = (
    assets[['id', 'criticality', 'processes']]
    .dropna(subset=['processes'])
    .assign(processus=lambda df: df['processes'].str.split(', '))
    .explode('processus')
)
asset_processes['poids_criticite_assets'] = asset_processes['criticality'].map(criticality_weight).fillna(1)

process_exposure = (
    asset_processes
    .groupby('processus')
    .agg(nombre_assets=('id', 'nunique'), score_exposition_assets=('poids_criticite_assets', 'sum'))
    .reset_index()
)

process_ranking = process_ranking.merge(process_exposure, on='processus', how='left')
process_ranking['score_priorite_total'] = (
    process_ranking['score_criticite_metier'] * 0.7
    + (process_ranking['score_exposition_assets'] / process_ranking['score_exposition_assets'].max() * 100) * 0.3
).round(1)

display(Markdown('## Classement des processus métier à protéger / surveiller en priorité'))
display(process_ranking.sort_values('rang_processus'))
```
